**1. Prompt the user for input. Score range should be from 0 to 10 (both inclusive).**

```python
Score=int(input("Enter the score between 0 and 10: "))

if 0<=Score<=10:
  print(f'Enterd valied score {Score}')
else:
  print("The Score must be between 0 and 10")
```

**2. Find the performance category based on the input score using the following criteria:**

* a. Above Average: Score greater than 7
* b. Average: Score between 4 and 7 (both inclusive)
* c. Below Average: Score less than 4

```python
Score=int(input("Enter the Score: "))

if 0<=Score<=10:
  if Score>7:
    print("Above Average:")
  elif 4<=Score<=7:
    print("Average")
  else:
   print("Below Average")
else:
  print("Enter score between 0 to 10")
```

**3. Output: Print the Performance category**

```python
Score=int(input("Enter the Score: "))

if 0<=Score<=10:
  if Score>7:
    print("Performance Category: Above Average:")
  elif 4<=Score<=7:
    print("Performance Category: Average")
  else:
   print("Performance Category: Below Average")
else:
  print("Enter score between 0 to 10")
```
